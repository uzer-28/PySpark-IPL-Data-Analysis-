```
from pyspark.sql.functions import *
```

```
#filter to include only valid delivaries (excluding wides and no balls for specific analyses)
ball_by_ball_df = ball_by_ball_df.filter((col('wides')==0) & (col('noballs')==0))
```

```
#Calculate the total and average runs scored in each match and innings
total_and_avg = ball_by_ball_df.groupby('match_id', 'innings_no').agg(
    sum(col('runs_scored')).alias('Total_Run'),
    avg(col('runs_scored')).alias('Average')
)
```

```
total_and_avg.display()
```

```
from pyspark.sql.window import Window
```

```
#Calculate the running total for each match in each over(window function)

window = Window.partitionBy('match_id').orderBy('over_id')
ball_by_ball_df = ball_by_ball_df.withColumn('running_total_sum',sum('runs_scored').over(window))
```

```
#Conditional Column: Flag for high impact balls (either a wicket or more than 6 runs including extras)
ball_by_ball_df = ball_by_ball_df.withColumn(
    'high_impact', 
    when((col('runs_scored')+col('extra_runs') > 6) | (col('bowler_wicket') == True), True).otherwise(False)
    )
```

```
ball_by_ball_df.limit(10).display()
```

# Analysing Match Dataframe

```
#Analysing day, month & year from match date for proper time based analysis

match_df = match_df.withColumn('year', year('match_date'))\
            .withColumn('month', month('match_date'))\
            .withColumn('day', dayofmonth('match_date'))
```

```
#Categorizing win margin into high, medium & low

match_df = match_df.withColumn(
    'win_margin_category',
    when((col('win_margin')>100), 'High')
    .when((col('win_margin')>50) & (col('win_margin')<=100), 'Medium')
    .otherwise('Low')
)
```

```
match_df.limit(5).display()
```

```
match_df = match_df.withColumn(
    'toss_match_winner',
    when((col('toss_winner')) == (col('match_winner')), 'Yes').otherwise('No')
)
```

```
top_scoring_batsmen_per_season = spark.sql("""
SELECT 
p.player_name,
m.season_year,
SUM(b.runs_scored) AS total_runs 
FROM ball_by_ball b
JOIN match m ON b.match_id = m.match_id   
JOIN player_match pm ON m.match_id = pm.match_id AND b.striker = pm.player_id     
JOIN player p ON p.player_id = pm.player_id
GROUP BY p.player_name, m.season_year
ORDER BY m.season_year, total_runs DESC
""")
```

```
top_scoring_batsmen_per_season.limit(40).display()
```

```
toss_impact_individual_matches = spark.sql("""
SELECT m.match_id, m.toss_winner, m.toss_name, m.match_winner,
       CASE WHEN m.toss_winner = m.match_winner THEN 'Won' ELSE 'Lost' END AS match_outcome
FROM match m
WHERE m.toss_name IS NOT NULL
ORDER BY m.match_id
""")
toss_impact_individual_matches.limit(20).display()
```
